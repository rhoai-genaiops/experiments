# 🧪 Synthetic Data Generation with SDG Hub

This notebook demonstrates how to generate synthetic training data for fine-tuning using **SDG Hub** - a modular Python framework for building synthetic data generation pipelines.

> ⚠️ **Workshop Scope:** Generating a full training dataset takes time and API costs we don't have in this workshop. This section explains the *process* and shows the tools. In the notebook, you'll generate a small sample to see how it works—not enough for real training, but enough to understand the pipeline.

## What We're Building

Training data for our Socratic math tutor. Each example shows:
- A student asking a math question
- The tutor responding with a guiding question (not an answer)

## The Pipeline

![SGD-pipeline.png](images/SGD-pipeline.png)

## SDG Hub Concepts

| Concept | Description |
|---------|-------------|
| **Blocks** | Individual processing units that transform data |
| **Flows** | YAML-configured pipelines that chain blocks together |
| **Registry** | Auto-discovery system for finding available flows and blocks |

SDG Hub flows are defined in YAML and orchestrate multiple steps:

1. **Chunk documents** — Break content into manageable pieces
2. **Generate student question** — LLM creates a realistic student question from the content
3. **Generate response** — LLM creates a response based on the content
4. **Format conversation** — Structure the Q&A pair for training

The flow calls your deployed LLM endpoint to generate data—this is an API call, not local inference, so it works fine from your CPU workbench.

## Step 3: Load and Configure a Flow

We'll use a document-grounded QA generation flow. This flow takes document content and generates question-answer pairs.

```python
# Load a QA generation flow
# In production, you'd use: flow = Flow.from_yaml(FlowRegistry.get_flow_path(flow_id))
# For this demo, we'll use a built-in flow or configure one manually

try:
    # Try to load a document QA flow
    qa_flows = FlowRegistry.search_flows(tag="question-generation")
    if qa_flows:
        flow_id = qa_flows[0]['id']  # Get the flow ID from the dict
        flow_path = FlowRegistry.get_flow_path(flow_id)
        flow = Flow.from_yaml(flow_path)
        print(f"✅ Loaded flow: {qa_flows[0]['name']} ({flow_id})")
    else:
        print("⚠️ No QA flows found, will use manual generation")
        flow = None
except Exception as e:
    print(f"⚠️ Could not load flow: {e}")
    print("   Will use manual generation instead")
    flow = None
```

```python
# If we have a flow, configure the model endpoint
if flow:
    # Configure to use the hosted Llama 3.2 3B endpoint
    # LiteLLM requires the "openai/" prefix for OpenAI-compatible endpoints
    MODEL_URL = "http://llama-32-predictor.ai501.svc.cluster.local:8080/v1"
    MODEL_NAME = "openai/llama32"  # LiteLLM provider prefix + model name
    API_KEY = "none"
    
    # Set model config for the flow
    flow.set_model_config(
        model=MODEL_NAME,
        api_base=MODEL_URL,
        api_key=API_KEY,
    )
    
    print(f"✅ Configured model:")
    print(f"   Provider: openai-compatible (LiteLLM)")
    print(f"   Model: {MODEL_NAME}")
    print(f"   Endpoint: {MODEL_URL}")
```

## Step 4: Prepare the Dataset

SDG Hub uses HuggingFace Datasets format. We'll prepare our document content for the flow.

```python
from datasets import Dataset
import pandas as pd
import re

def chunk_by_sections(markdown: str) -> list[dict]:
    """Split markdown into sections based on headers."""
    # Split by level 2 headers (##)
    sections = re.split(r'\n##\s+', markdown)
    
    chunks = []
    for i, section in enumerate(sections):
        if section.strip():
            # Extract title from first line
            lines = section.strip().split('\n')
            title = lines[0].strip('#').strip() if lines else f"Section {i}"
            content = '\n'.join(lines[1:]).strip() if len(lines) > 1 else section.strip()
            
            if len(content) > 100:  # Only include substantial sections
                # Create a simple outline from the content
                outline_lines = [l.strip() for l in content.split('\n') if l.strip().startswith(('-', '*', '•'))]
                outline = '\n'.join(outline_lines[:5]) if outline_lines else f"Section about {title}"
                
                chunks.append({
                    'document': content,
                    'document_outline': outline,
                    'title': title,
                    'domain': 'Mathematics Education',
                    # In-context learning examples for Socratic tutoring
                    'icl_document': 'A student learning to solve linear equations needs guidance, not answers.',
                    'icl_query_1': 'How do I solve 2x + 5 = 13?',
                    'icl_response_1': 'What operation would help you isolate the variable x?',
                    'icl_query_2': 'What is the derivative of x squared?',
                    'icl_response_2': 'What rule do you know for finding derivatives of power functions?',
                    'icl_query_3': 'I dont understand quadratic equations.',
                    'icl_response_3': 'What form does a quadratic equation typically take? Can you write one?',
                })
    
    return chunks

# Create chunks from our markdown
chunks = chunk_by_sections(markdown_content)
print(f"✅ Created {len(chunks)} document chunks")

# Show first chunk
if chunks:
    print(f"\n--- First chunk ---")
    print(f"Title: {chunks[0]['title']}")
    print(f"Document preview: {chunks[0]['document'][:150]}...")
    print(f"Outline: {chunks[0]['document_outline'][:100]}...")
```

```python
# Create a dataset for the flow
if flow:
    # Get the schema expected by the flow (returns empty pandas DataFrame)
    schema_df = flow.get_dataset_schema()
    required_cols = list(schema_df.columns)
    print(f"Flow expects columns: {required_cols}")
    
    # Create a DataFrame with our chunks
    input_df = pd.DataFrame(chunks)
    
    # Verify our data has the required columns
    missing_cols = set(required_cols) - set(input_df.columns)
    if missing_cols:
        print(f"⚠️ Missing columns: {missing_cols}")
        print("   Adding placeholder values for missing columns...")
        for col in missing_cols:
            input_df[col] = f"[placeholder for {col}]"
    
    print("✅ All required columns present")
    input_dataset = input_df
else:
    # Create HuggingFace Dataset directly (for manual generation)
    input_dataset = Dataset.from_list(chunks)

print(f"\n✅ Dataset ready: {len(input_dataset)} examples")
cols = list(input_dataset.columns) if hasattr(input_dataset, 'columns') else input_dataset.column_names
print(f"   Columns: {cols}")
```

## Step 5: Generate Synthetic Data

Now we run the generation pipeline. For production, you'd use the full SDG Hub flow. Here we'll demonstrate both approaches.

We will also limit the number of training examples we generate here to 2 samples, as they take a fair bit of time to generate: 

| What You Need | Workshop Demo | Production |
|---------------|---------------|------------|
| Training examples | 2 samples | 500-5000 |
| Generation time | ~2-5 minutes | Hours to days |
| API calls | Minimal | Thousands |

```python
# If we have a flow, do a dry run first
if flow:
    print("Running dry run to estimate time...")
    dry_result = flow.dry_run(input_dataset, sample_size=1)
    print(f"\nDry run complete!")
    print(f"Estimated time for full generation: {dry_result.get('estimated_time', 'N/A')}")
```

```python
# Generate synthetic Q&A pairs (limited sample for demo)
SAMPLE_SIZE = 2  # Limit to 2 samples for faster generation

if flow:
    print(f"Generating synthetic data with SDG Hub (limited to {SAMPLE_SIZE} samples)...")
    print("(This may take a few minutes)\n")
    
    # Limit input dataset to speed up generation
    limited_dataset = input_dataset.head(SAMPLE_SIZE) if hasattr(input_dataset, 'head') else input_dataset[:SAMPLE_SIZE]
    
    result = flow.generate(limited_dataset)
    
    # Extract generated data and convert to training format
    questions = result['question'].tolist() if hasattr(result['question'], 'tolist') else list(result['question'])
    answers = result['response'].tolist() if hasattr(result['response'], 'tolist') else list(result['response'])
    
    print(f"✅ Generated {len(questions)} Q&A pairs")
    
    # Convert to training conversation format (same as manual generation)
    training_data = []
    for q, a in zip(questions, answers):
        training_data.append({
            "messages": [
                {"role": "user", "content": q},
                {"role": "assistant", "content": a}
            ]
        })
    
    # Show examples
    for i in range(min(3, len(training_data))):
        print(f"\n--- Example {i+1} ---")
        print(f"Q: {training_data[i]['messages'][0]['content'][:100]}...")
        print(f"A: {training_data[i]['messages'][1]['content'][:100]}...")
else:
    print("⚠️ No SDG Hub flow available, will use manual generation in next cell")
    training_data = None  # Will be populated by manual generation
```

## Step 6: Quality Filtering

Not all generated data is good. We filter to ensure responses actually follow the Socratic method.

In production SDG Hub flows, this filtering is built into the pipeline with quality scores (faithfulness, relevancy).

```python
def is_socratic(response: str) -> bool:
    """Check if response follows Socratic method."""
    
    # Red flags - giving answers directly (be more specific to avoid false positives)
    answer_patterns = [
        "the answer is",
        "the solution is",
        "therefore, x =",
        "so x =",
    ]
    
    # Green flags - asking guiding questions
    question_patterns = [
        "what",
        "how",
        "why",
        "can you",
        "have you",
        "do you",
        "?"
    ]
    
    response_lower = response.lower()
    
    # Reject if it gives answers
    for pattern in answer_patterns:
        if pattern in response_lower:
            return False
    
    # Accept if it asks questions
    for pattern in question_patterns:
        if pattern in response_lower:
            return True
    
    return False

# First, let's see what the responses look like before filtering
print("Sample responses before filtering:")
for i, conv in enumerate(training_data[:5]):
    response = conv["messages"][1]["content"]
    is_ok = is_socratic(response)
    status = "✅" if is_ok else "❌"
    print(f"\n{status} Example {i+1}:")
    print(f"   Q: {conv['messages'][0]['content'][:80]}...")
    print(f"   A: {response[:150]}...")

# Apply filter
filtered_data = [
    conv for conv in training_data
    if is_socratic(conv["messages"][1]["content"])
]

print(f"\n" + "="*60)
print(f"Quality filtering results:")
print(f"  Original: {len(training_data)} conversations")
print(f"  After filter: {len(filtered_data)} conversations")
print(f"  Kept: {len(filtered_data)/len(training_data)*100:.0f}%" if training_data else "N/A")

# If too few passed, warn and offer to skip filtering
if len(filtered_data) < 10 and len(training_data) > 10:
    print(f"\n⚠️  Very few responses passed the Socratic filter.")
    print(f"   The SDG Hub flow generates informative responses, not Socratic ones.")
    print(f"   For this demo, we'll use the unfiltered data.")
    filtered_data = training_data
    print(f"   Using all {len(filtered_data)} conversations.")
```

```python
# Test the filter on some examples
test_responses = [
    "The answer is x = 4",  # Should fail
    "What operation would help you isolate x?",  # Should pass
    "You get 8 when you subtract 5",  # Should fail
    "What do you notice about the left side?"  # Should pass
]

print("Quality filter test:")
for response in test_responses:
    result = "✅ PASS" if is_socratic(response) else "❌ FAIL"
    print(f"{result}: {response[:50]}...")
```

## Step 7: Save Training Data

Training data is saved in JSONL format (one JSON object per line). This is the format expected by the training notebook.

```python
import json

# Save as JSONL
output_path = "socratic-training-data.jsonl"

with open(output_path, "w") as f:
    for conversation in filtered_data:
        f.write(json.dumps(conversation) + "\n")

print(f"✅ Saved {len(filtered_data)} examples to {output_path}")
print(f"   File size: {os.path.getsize(output_path) / 1024:.1f} KB")

# Show the file contents
print(f"\n--- File contents (first 3 lines) ---")
with open(output_path, "r") as f:
    for i, line in enumerate(f):
        if i >= 3:
            break
        data = json.loads(line)
        print(f"\nExample {i+1}:")
        print(f"  Student: {data['messages'][0]['content'][:60]}...")
        print(f"  Tutor: {data['messages'][1]['content'][:60]}...")
```
